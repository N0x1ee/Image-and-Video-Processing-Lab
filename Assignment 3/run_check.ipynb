**1. WAP to draw a border around the input image (pad the input image).**

Input the width of the border (in terms of pixel count) and the desired colour of the border from the user. Try this for both binary and grayscale image.

The colour is entered as R, G and B. A grayscale image has a single channel, so the colour is converted to its equivalent gray level, and a binary image can hold only 0 or 255, so that gray level is rounded to whichever is nearer.

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("Me.png", 0)

plt.imshow(image, cmap="gray")
plt.axis("off")
plt.show()
```

```python
width = int(_input("Enter border width: "))

r = int(_input("Enter R (0-255): "))
g = int(_input("Enter G (0-255): "))
b = int(_input("Enter B (0-255): "))

gray_color = int(round(0.299*r + 0.587*g + 0.114*b))
binary_color = 255 if gray_color >= 128 else 0

print("Gray level of the colour:", gray_color)
print("Value used for binary  :", binary_color)
```

```python
gray_border = cv2.copyMakeBorder(image, width, width, width, width,
                                 cv2.BORDER_CONSTANT, value=gray_color)

_, binary = cv2.threshold(image, 127, 255, cv2.THRESH_BINARY)

binary_border = cv2.copyMakeBorder(binary, width, width, width, width,
                                   cv2.BORDER_CONSTANT, value=binary_color)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(gray_border, cmap="gray")
plt.title("Grayscale Border")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(binary_border, cmap="gray")
plt.title("Binary Border")
plt.axis("off")

plt.show()
```

Try this for both binary and grayscale image.

```python
gray_complement = cv2.bitwise_not(image)
binary_complement = cv2.bitwise_not(binary)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(gray_complement, cmap="gray")
plt.title("Grayscale Complement")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(binary_complement, cmap="gray")
plt.title("Binary Complement")
plt.axis("off")

plt.show()
```
